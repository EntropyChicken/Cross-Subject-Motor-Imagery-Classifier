Data itself goes to a sibling directory to the repo, so as not to clutter the repo

```python
from mne.datasets import eegbci
```

```python
def download(subject_ids: list[int], target_runs: list[int]):
    all_file_paths = {}

    for subject in subject_ids:
        print(f"Checking/fetching Subject {subject:03d}...")
        paths = eegbci.load_data(subjects=subject, runs=target_runs, update_path=False)
        all_file_paths[subject] = paths

    print("\nDownload complete. Total files managed:", sum(len(p) for p in all_file_paths.values()))
```

```python
download(list(range(1, 11)), [3, 4, 7, 8, 11, 12]) # main runs for subjects 1-10
```
